Dane zostały oczyszczone i przygotowane do dalszych obliczeń. Na etapie weryfikacji poprawności danych pozbyłem się 660 wierszy, z powodu duplikatów. Dodatkowo NaN w kolumnie statsbomb_xg zostało zastąpione 0.

```python
wynik = {'Polonia Bytom': 0, 'Pogoń Grodzisk Mazowiecki': 0}

statystyki_xg = {
    'Polonia Bytom': {
        'Prowadzenie': {'xGF': 0.0, 'xGA': 0.0},
        'Remis': {'xGF': 0.0, 'xGA': 0.0},
        'Przegrana': {'xGF': 0.0, 'xGA': 0.0}
    },
    'Pogoń Grodzisk Mazowiecki': {
        'Prowadzenie': {'xGF': 0.0, 'xGA': 0.0},
        'Remis': {'xGF': 0.0, 'xGA': 0.0},
        'Przegrana': {'xGF': 0.0, 'xGA': 0.0}
    }
}

for _, row in dane.iterrows():

    if wynik['Polonia Bytom'] > wynik['Pogoń Grodzisk Mazowiecki']:
        stan_polonia = 'Prowadzenie'
        stan_pogon = 'Przegrana'
    elif wynik['Polonia Bytom'] < wynik['Pogoń Grodzisk Mazowiecki']:
        stan_polonia = 'Przegrana'
        stan_pogon = 'Prowadzenie'
    else:
        stan_polonia = 'Remis'
        stan_pogon = 'Remis'

    if row['event_type_name'] == 'Shot':
        xg = row['statsbomb_xg']
        druzyna = row['team_name']

        if druzyna == 'Polonia Bytom':
            statystyki_xg['Polonia Bytom'][stan_polonia]['xGF'] += xg
            statystyki_xg['Pogoń Grodzisk Mazowiecki'][stan_pogon]['xGA'] += xg
        elif druzyna == 'Pogoń Grodzisk Mazowiecki':
            statystyki_xg['Pogoń Grodzisk Mazowiecki'][stan_pogon]['xGF'] += xg
            statystyki_xg['Polonia Bytom'][stan_polonia]['xGA'] += xg

    if row['outcome_name'] == 'Goal':
        wynik[row['team_name']] += 1

dane_do_tabeli = []

for druzyna in statystyki_xg:
    for stan in ['Prowadzenie', 'Remis', 'Przegrana']:
        xgf = statystyki_xg[druzyna][stan]['xGF']
        xga = statystyki_xg[druzyna][stan]['xGA']

        dane_do_tabeli.append({
            'Drużyna': druzyna,
            'Stan meczu': stan,
            'xGF': round(xgf, 3),
            'xGA': round(xga, 3),
            'Różnica xG': round(xgf - xga, 3)
        })
tabela_wynikow = pd.DataFrame(dane_do_tabeli)
tabela_wynikow
```

Powyższy fragment kodu oblicza bilans oczekiwanych goli (xG) w zależności od stanu meczu (Prowadzenie, Remis, Przegrana).

Dynamiczny wynik -  Pętla na bieżąco śledzi wynik, ustalając kontekst na boisku przed każdym zdarzeniem.

Kategoryzacja strzałów - Wartość statsbomb_xg powiększa pulę wykreowanego (xGF) i dopuszczonego (xGA) zagrożenia dla odpowiednich drużyn w bieżącym stanie gry.

Chronologia zdarzeń -  Bramki zmieniają wynik dopiero po zaksięgowaniu strzału. Dzięki temu gol dający prowadzenie poprawnie przypisuje się do statystyk przy "Remisie".

Zestawienie końcowe: Agregowane dane trafiają do tabeli wraz z obliczoną różnicą xG.

```python
tabela_wynikow.to_csv('wyniki_xg_pogon.csv', index=False, sep=';', decimal=',')
```
